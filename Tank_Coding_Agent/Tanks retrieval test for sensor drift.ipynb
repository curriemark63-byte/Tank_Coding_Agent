```python
import numpy as np
import matplotlib.pyplot as plt

class SensorDriftSimulator:
    def __init__(self, initial_value=0.0, bias_drift_rate=0.01, noise_level=0.05, time_steps=1000):
        self.initial_value = initial_value
        self.bias_drift_rate = bias_drift_rate
        self.noise_level = noise_level
        self.time_steps = time_steps
        self.time_array = np.linspace(0, time_steps, time_steps)
        self.bias_values = np.zeros(time_steps)
        self.sensor_values = np.zeros(time_steps)

    def simulate_sensor_drift(self):
        current_value = self.initial_value
        for t in range(self.time_steps):
            bias_drift = self.bias_drift_rate * current_value
            noise = self.noise_level * np.random.randn()
            current_value += bias_drift + noise
            self.bias_values[t] = bias_drift
            self.sensor_values[t] = current_value

    def plot_bias_drift(self):
        plt.figure(figsize=(10, 6))
        plt.plot(self.time_array, self.bias_values, label='Bias Drift')
        plt.xlabel('Time Steps')
        plt.ylabel('Bias Drift')
        plt.title('Sensor Bias Drift Over Time')
        plt.legend()
        plt.grid(True)
        plt.show()

if __name__ == "__main__":
    simulator = SensorDriftSimulator(initial_value=0.5, bias_drift_rate=0.001, noise_level=0.02, time_steps=1000)
    simulator.simulate_sensor_drift()
    simulator.plot_bias_drift()
```
